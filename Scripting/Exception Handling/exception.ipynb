# Exception Handling
- An exception is a runtime error that stopes your program usless handled.

```python
x = 10/0
```

```python
try:
  d = int(input("Enter a number: "))
  x = 10/d
  print("Result:", x)
except ZeroDivisionError:
  print("Denominator cannot be zero")
```

```python
try:
  a = int(input("Enter a numerator: "))
  b = int(input("Enter a denominator: "))
  x = a / b
  print("Result:", x)
except ZeroDivisionError:
  print("Denominator cannot be zero")
except ValueError:
  print("Invalid input, please enter a valid integer")
```

```python
try:
  birth_year = int(input("Enter your birth year: "))
  current_year = int(input("Enter the current year: "))
  age = current_year - birth_year
  if age < 0:
    raise ValueError("Future birth year is not allowed")
  print("Your age is:", age)
except ValueError as e:
  print("Error:", e)
```

# Using else and finally

```python
try:
  with open("logs.txt", "r") as file:
    content = file.read()
except FileNotFoundError:
  print("File not found. Please check the file path and try again.")
else:
  print(f"File content:\n{content}")
finally:
  print("\n----Task Completed----")
```

```python
def valid_data_check(line):
  parts = line.split(",")
  for part in parts:
    part = part.strip()
    if "-" not in part:
      raise ValueError(f"Invalid data format: {part}")
    
    name, mks = part.split(" - ",1)
    name = name.strip()
    mks = mks.strip()

    if "**FAILD**" in mks:
      mks = mks.replace("**FAILD**", "").strip()
    
    if not mks.isdigit():
      raise ValueError(f"Invalid marks for {name}: {mks}")
    
    print(f"Valid Entry: {name} - {mks}")

try:
  with open("data.txt", "r") as file:
    content = file.read().strip()

    if not content:
      raise ValueError("File is empty")
    else:
      valid_data_check(content)

except FileNotFoundError:
  print("File not found. Please check the file path and try again.")
except ValueError as ve:
  print("Error:", ve)
except Exception as e:
  print("An unexpected error occurred:", e)
finally:
  print("\n----Data Validation Completed----")

```

## Custum Expections

- Python has many built-in exceptions (ValueError, TypeError, ZeroDivisionError, etc.), but sometimes you want to raise your own special errors with a custom name and message.
- This makes your code easier to debug and more meaningful.
- You always inherit from the built-in Exception class.
- You can add a custom message, or even custom logic if needed.
- The docstring explains what this exception means (useful in larger projects)

```python
class InvalidInputError(Exception):
  """
  Custom exception for invalid input errors.
  """
  pass

try:
  mark = "absc"
  print(mark.isdigit())
  if not mark.isdigit():
    raise InvalidInputError("Invalid input: marks should be a number")
  print("Marks are valid:", mark)
  
except InvalidInputError as e:
  print("Error:", e)
```
