# Automation
Automation means writing scripts that do repetitive tasks automatically, without needing manual input everytime.

### Python is perfect for automation because of its:
- clean syntax.
- cross-platform support.
- powerfull libraries like os, shutil, datetime, pandas, selenium, etc.


## Real world automations ex.)

Task - automation idea
- Sorting download files - Move PDFs, ZIPs, ss to folders
- Renaming hundreds of files - add timestamos, remove unwanted names
- Sending reminder emails - use gmail or whatsapp automation
- auto-filling web forms - Using Selenium or pyautogui
- Logging student attendence - Daily log written to file.
