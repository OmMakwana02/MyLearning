## Opening using "with open(...)"
- We use this for safety and auto-closing.

```python
with open('logs.txt', 'r') as file2:
  for line in file2:
    print(line.strip())
```

```python
with open('logs.txt', 'r') as file2:
  i = 1
  for line in file2:
    print(f"Line {i}: {line.strip()}")
    i += 1

  
```

```python
with open('logs.txt', 'r') as file2:
  count, j = 0, 1
  for line in file2:
    count = len(line.strip().split())
    print(f"line {j} ({count} words): {line.strip()}")
    j += 1
```

## Writing in a file.
- we use "w" for writing.
- this overwrites if the file exists

```python
with open("logs.txt", "w") as file3:
  file3.write("[INFO] This is a new log entry.\n")
  file3.write("[ERROR] An error occurred.\n")
  file3.write("[ERROR4] Data cannot be loaded.\n")
  file3.write("[ERROR 404] Content not found.\n")
  file3.write("[TASK DONE] Succes.\n")

with open("logs.txt", "r") as f1:
  cont = f1.read()
print("Content of logs.txt after writing new entries:")
print(cont) 
```

```python
with open("daily_logs.txt", "a") as df:
  while True:
    i = int(input("Choose: 1.) Add daily log or 2.) Read the logs. 3.) Exit: "))
    if i == 1:
      daily_log = input("Enter the log: ")
      df.write(f"{daily_log}\n")
    elif i == 2:
      with open("daily_logs.txt", "r") as df1:
        rd = df1.read()
        print("Daily Logs:")
        print(rd)
    elif i == 3:
      print("Exiting the program.")
      break
    else:
      print("Invalid choice. Please choose 1 or 2 or 3.")
```

## Now reading csv file

```python
import csv
```

```python
with open("data.csv", "r") as file4:
  df = csv.reader(file4)
  info = None
  for row in df:
    if info is None:
      info = row
    else:
      print(f"{info[0]}: {row[0]}, {info[1]}: {row[1]}, {info[2]}: {row[2]}")
```
