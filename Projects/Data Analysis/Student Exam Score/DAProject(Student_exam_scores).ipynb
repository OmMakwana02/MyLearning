##--> As columns "Unnamed: 0" has no use to us, we are droping it.

```python
df = df.drop("Unnamed: 0", axis = 1)
# Here "axis = 1" means we have to search/look into columns.
```

```python
df.head()
```

```python
plt.figure(figsize = (5,6))
gc = sns.countplot(data = df, x = "Gender")
# To label the count in a plot we use this....
gc.bar_label(gc.containers[0])
plt.title("Count of Genders")
plt.show()
```

## --> From th eabove chart we have analysed that the no. of female students are more as compared to male students.

```python
# Groupby function: used to split dfs into groups based on values in one or more columns.
# agg() function: used to perform calculations on the groups created by groupby().
grp1 = df.groupby("ParentEduc").agg({"MathScore":"mean", "ReadingScore":"mean", "WritingScore":"mean"})
print(grp1)
```

```python
sns.heatmap(grp1, annot = True) # annot: wether to display values of heatmap cells directly on the plot.
plt.title("Relation between Parent's Education and Student's Marks.")
plt.show()
```

##--> From the above plot we can say that the parents with higher education like master's degree, bachelor's degree, etc.

```python
df.head()
```

```python
grp2 = df.groupby("ParentMaritalStatus").agg({"MathScore":"mean", "ReadingScore":"mean", "WritingScore":"mean"})
print(grp2)
```

```python
sns.heatmap(grp2, annot = True)
plt.title("Relation between Parent's Marital Status and Student's Marks.")
plt.show()
```

## --> From the above plot we can see that the there is negligible difference between marks of students, whose parents are either married, divorced, single or widowed...

```python
grp3 = df.groupby("LunchType").agg({"MathScore":"mean", "ReadingScore":"mean", "WritingScore":"mean"})
print(grp3)
```

```python
sns.heatmap(grp3, annot = True)
plt.title("Relation between Lunch Type and Student's Marks.")
plt.show()
```

##--> Here we can see that the students who consume standard food have good potential in scoring good marks as compared to the students who consume free/reduced food.

```python
grp4 = df.groupby("WklyStudyHours").agg({"MathScore":"mean", "ReadingScore":"mean", "WritingScore":"mean"})
print(grp4)
```

```python
sns.heatmap(grp4, annot = True)
plt.title("Relation between Weekly Study Hours and Student's Marks.")
plt.show()
```

##--> The above plot shows that the the students how study 5-10 hours a week or more than 10 hours tend to score little more marks in the exam as compared to students how study less than 5 hours a week.

```python
grp5 = df.groupby("TestPrep").agg({"MathScore":"mean", "ReadingScore":"mean", "WritingScore":"mean"})
print(grp5)
```

##--> This shows that people who have no test preparation score less marks in the exam as compared to the people who have completed the test preparation.

```python
print(df["EthnicGroup"].unique())
# unique(): returns an array containing only the unique values from that column.
# In other words, it finds all the distinct ethnic groups present in your data without repetition.
```

```python
ag = sns.countplot(data = df, x = "EthnicGroup")
ag.bar_label(ag.containers[0])
plt.title("Count of Ethnic Groups")
plt.show()
```

```python
grpA = df.loc[(df["EthnicGroup"] == 'group A')].count()
print("Group A:\n",grpA)
grpB = df.loc[(df["EthnicGroup"] == 'group B')].count()
print("Group B:\n",grpB)
grpC = df.loc[(df["EthnicGroup"] == 'group C')].count()
print("Group C:\n",grpC)
grpD = df.loc[(df["EthnicGroup"] == 'group D')].count()
print("Group D:\n",grpD)
grpE = df.loc[(df["EthnicGroup"] == 'group E')].count()
print("Group E:\n",grpE)
# loc(): it is primarily label-based indexing for selection by label.
# This means you use row and column labels to access specific data within a Pandas DataFrame (df).
```

```python
l = ["group A", "group B", "group C", "group D", "group E"]
list = [grpA["EthnicGroup"], grpB["EthnicGroup"], grpC["EthnicGroup"], grpD["EthnicGroup"], grpE["EthnicGroup"]]
plt.pie(list, labels = l,autopct = "%1.2f%%" ,startangle=720)
plt.title("Percentage of Ethnic Groups")
plt.show()
# autopct: The autopct parameter enables you to display the percentage value of each slice (or wedge) directly on the pie chart.
# startangle: The startangle parameter controls the angle (in degrees) at which the first slice (or wedge) of the pie chart is drawn.
```
