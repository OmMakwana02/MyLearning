## 8.1 Import modules and Packages in Python
In Python Modules and Packages help organize and reuse code. Here's a comprehensive guide on how to import them.

```python
# Importing a packades
import math
```

```python
math.sqrt(17)
```

```python
# Method 2 to import functions from a package
# Syntax: from package import function1, function2

from math import sqrt, pi
print(sqrt(50055))
print(sqrt(25))
print(pi)
```

```python
import numpy as np
# np is just an alias
```

```python
# To import every module in any package we use *
from math import *
# After this we can directly use the the functions in the packages...
print(sqrt(64))
print(sinh(56))
print(sinh(pi/2))
```

### Steps to create our own custom package.
S1> Make a folder name package

S2> In it make a "__ init __.py" file.

(no space between the __ and init)

(__ init __.py is a special file used in python to define packages and initialixe their namespaces)

S3> Make packages py file and write functions you want in that package.


```python
# importing our own custom made package
from package.maths import *

addition(8,9)
```

```python
subtraction(8,9)
```

We can also create a subpackage in a package...

S1> make a subpackage folder in the package folder, and follow the same steps as mentioned above.

Syntax to access that package:

from package.subpackage.package_name import *

```python
# Data Serialization

import json

data = {'name':'Rachel', 'age':17}

json_str = json.dumps(data)
# this will convert the dictonary to a string. 
print(json_str)
print(type(json_str))

# To convert back into the dictonary
parsed_data = json.loads(json_str)
print(parsed_data)
print(type(parsed_data))
```

```python
# Datetime

from datetime import datetime, timedelta

now = datetime.now()
print(now)

yesterday = now-timedelta(days=1)
print(yesterday)
```

```python
# Regular Expression
import re

pattern = r'\d+'
text = 'there are total of 145 apples and 12 bananas'
match = re.search(pattern, text)
print(match.group())
```
