# Section 9: File Handling in Python

File handling is a crucial part of programming language. Python provides built-in functions and methods to read from and write to files, both text and binary. This includes the basis of file handling, including reading and writing text files and binary files. 

## Text Files

```python
# Read a whole file

with open('example.txt','r') as file:
    content = file.read()
    print(content)
```

```python
# Read a file, line by line

with open('example.txt', 'r') as file:
    for line in file:
        print(line)
```

```python
with open('example.txt', 'r') as file: # 'r' for read mode
    for line in file:
        print(line.strip())
# This .strip() removes the newline character.
```

```python
# Writing a file(overwriting)

with open('example.txt', 'w') as file: # 'w' for write mode
    file.write('We Dont know if it really exists or not.\n')
    file.write('But we assume that it is real and we do our dutes as a friend.')

# This will over write everything in the file....the previous content is gone(overwritten by the above write function).
```

```python
 # Writing in a file(without overwriting)

with open('example.txt', 'a') as file: # 'a' for append mode
    file.write('We Dont know if it really exists or not.\n')
    file.write('But we assume that it is real and we do our dutes as a friend.\n')
```

```python
# Writing a list of lines to a file.

lines = ['As we try to be the real friend in the mean time we also expect the other person to do same as we do.\n', 'But in some cases its not true, we become the perfect friend but the person whome we expect to be the same real friend, does not fullfil our ecpectations.\n', 'Then we suffer from pain...we tell ourself, that everything will be okay. \n', 'But nothing is okay, because we give our 100% to this friendship, but when we dont get the same 100%, we feel betrayed, hurt and we suffer through immense amount of pian.\n']

with open('example.txt', 'a') as file:
    file.writelines(lines)
```

## Binary Files


```python
# Writing in Binary files
data = b'\x00\x01\x02\x03\x04'
with open('example.bin', 'wb') as file: # 'wb' for write bite
    file.write(data) 
```

```python
# Reading binary files
with open('example.bin', 'rb') as file: # 'rb' for read bite
    data = file.read()
    print(data)
```

```python
# Read content from a source text file and write it to a destination text file.
# copying a text file
with open('example.txt','r') as source_file:
    content = source_file.read()

with open('destnation.txt', 'w') as destination_file:
    destination_file.write(content)
```
