### Now we will see how we can write this code in just a single line.

--> So list comprehension has 2 important things,  

Basic Syntax  [expression for item in iterable]

Syntax with conditional statement [expression for item in iterable if statement]

Nested Comphrension [ecpression for item1 in iterable1 for item2 in iterable2]

```python
# Basic Syntax
list = [i**2  for i in range(10)]
print(list)
```

```python
li = [1,2,3,44]
li.reverse()
print(li)
```

```python
# Syntax with conditional statement
list1 = [i**2 for i in range(10) if i%2==0]
print(list1)
```

```python
# Nested comphrension
lst1 = [1,2,3,4,5,6]
lst2 = ['a', 'b', 'c', 'd']

pair = [[i, j] for i in lst1 for j in lst2]
print(pair)
```

### Mathematical Operationsnon set

1. union

-> set1.union(set2)

2. Intersection

-> set1.intersection(set2)

3. Intersection Update - This function will give the intersection of the set and it eill update the set 1, it will give the common elements in the set and update them to set 1

-> set1.intersection_update(set2)

4. Difference

-> set1.difference(set2)

5. Symmetric Difference - From set 1 and set2 all the common elements are removed and only unique elements are left behind.

-> set1.symmetric_difference(set2)

```python
# Set Methods

set1={1,2,3}
set2={4,5,6}
set3={2,3}
## is subset
print(set1.issubset(set2))
print(set3.issubset(set1))

## superset
print(set1.issuperset(set2))
print(set1.issuperset(set3))
```

## Iterating Over a Dictonary
### You can use loops to iterate over dictionaries, kwys, values or items. 

```python
# Iterating over keys
for keys in student.keys():
    print(keys)
```

```python
# Iterating over values
for values in student.values():
    print(values)
```

```python
# Iterate over key value pairs 
for key, value in student.items():
    print(f"{key}:{value}")
```

```python
# Nested Dictionaries
studentss={
    "stud1": {"name":"Srishti","age":20},
    "stud2": {"name":"Om","age":21},
    "Stud3": {"name":"Falguni","age":22}
}
print(studentss)
# The data is stored like this, in any non-sequal data base...eg) MongoDB.
```

```python
# Accessing Nested Dictionarie's Elements
print(studentss["stud2"]["name"])
print(studentss["stud2"]["age"])
print(studentss["stud2"])

```

```python
# Itersating over Nested Dictionaries.
for student_id, student_info in studentss.items():
    print(f"{student_id}:{student_info}")
    for keys, values in student_info.items():
        print(f"{keys}:{values}")
```

```python
# Practical Examples
## Q. Use a dictionary to count the frequency of elements in list.

nums=[1,2,3,4,5,6,7,8,9,2,3,2,32,4,1,2,7,5,4,6,2,4,3,4,3,0,3,7,6,3,1]

count = {}
for keyz in nums:
    if keyz in count:
        count[keyz] += 1
    else:
        count[keyz] = 1

print(count)
```

```python
# Merge 2 dictionaries into one
dict1 = {"a":1, "b":2}
dict2 = {"b":3, "c":4}

merged_dictionary = {**dict1,**dict2}
# Here if im giving two **, this will take any no. of key value pair and just append them to the dictionary. This is called as a key-word argument.
print(merged_dictionary)
```

### Real World Examples Using List in Python
List are one of the most commonly used data structurs in Python, thanks to their versatility and ease of use. Here are several practical examples that illustrate their use in real world scenarios

```python
# Example 1: Manage a To-Do List.

to_do_list = []

# Add a new task to the list
n=0
while n != 3:
    n=int(input("Choose: 1. Add Task, 2. To remove task or 3. Exit"))
    if n == 1:
        to_do_list.append(input("Enter the task:"))
        print(to_do_list)
    elif n == 2:
        task_to_remove = input("Enter the tast to be removed.")
        flag = 0jmn 1
        for task in to_do_list: 
            if task == task_to_remove:
                to_do_list.remove(task)
                flag = 1
                print(f"Task {task} removed.")
                break
        if flag == 0:
            print(f"Task {task_to_remove} not found.")
    else:
        break

print(f"Gentle remainder, Tasks remaining:")
for tasks in to_do_list:
    print(f"-{tasks}")
```
