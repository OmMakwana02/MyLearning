# 13.5 Data Visualization with Matplotlib
- Matplotlib is a powerful plotting library for Python that enables the creation of static, animated and interactive visualizations. 
- It is widely used for data visualization in data science and analytics. In this session, we will cover the basics of Matplotlib, including creating various types of plots and customizing them.

```python
import matplotlib.pyplot as plt
```

```python
x=[1,2,3,4,5]
y=[1,4,9,16,25]

plt.plot(x,y)
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.title('Basic Line Plot')
plt.show()
```

```python
x=[1,2,3,4,5]
y=[1,4,9,16,25]

# Creating a customize line plot
plt.plot(x,y, color='red', linestyle='-') 
```

```python
plt.plot(x,y, color='red', linestyle='--', marker='o', linewidth=3, markersize=8)
plt.grid(True) 
```

```python
plt.plot(x,y, color='red', linestyle='-.', marker='*') 
```

```python
plt.plot(x,y, color='red', linestyle=':') 
```

### Multiple Plots

```python
## Sample Data
x = [1,2,3,4,5]
y1 = [i**2 for i in range(1,6)]
y2 = [j**3 for j in range(1,6)]

plt.figure(figsize=(9,5))
# plt.subplot(rows, columns, position)
plt.subplot(1,2, 1)
plt.plot(x,y1, color='green')
plt.title('Plot1')

plt.subplot(1,2, 2)
plt.plot(y1,x, color='green')
plt.title('Plot2')
```

```python
x = [1,2,3,4,5]
y1 = [i**2 for i in range(1,6)]
y2 = [j**3 for j in range(1,6)]

plt.figure(figsize=(15,12))
# plt.subplot(rows, columns, position)
plt.subplot(3,2, 1)
plt.plot(x,y1, color='green')
plt.title('Plot1')

plt.subplot(3,2, 2)
plt.plot(y1,x, color='green')
plt.title('Plot2')

plt.subplot(3,2, 3)
plt.plot(x,y2, color='blue')
plt.title('Plot3')

plt.subplot(3,2, 4)
plt.plot(y2,x, color='blue')
plt.title('Plot4')
```

### Bar Plot

```python
categories=['A','B','C','D','E']
values=[5,6,4,7,8]
```

```python
# Creating a Bar Plot
plt.bar(categories, values, color='purple')
plt.xlabel('Categories')
plt.ylabel('Values')
plt.title('Bar Plot')
```

### Histogram
Histogram are used to represent the distribution of a dataset. They divide the data into bins and count the number of data points in each bin.

```python
# Sample Data
data = [1,2,2,3,3,3,5,4,4,5,4,3,4,3,5,5,5,5,5]

# Create a Histogram
plt.hist(data, bins=5, edgecolor='black')
```

### Scatter Plot

```python
x=[1,2,3,4,5]
y=[3,4,5,6,7]

plt.scatter(x,y, color='blue')
```

```python
plt.scatter(x,y, color='blue', marker='*') # Markers = * / x / o
```

### Pie Chart

```python
labels=['A','B','C','D']
sizes=[30,20,40,10] 
coloors=['gold','yellowgreen','red','pink']
explode=(0.2,0,0,0)
```

```python
plt.pie(sizes, labels=labels, colors=coloors, autopct="%1.1f%%")
```

```python
plt.pie(sizes,explode=explode, labels=labels, colors=coloors, autopct="%1.1f%%")
```

```python
total_product=sales_df.groupby('Category')['Amount'].sum()
print(total_product)
```

```python
# Plot sales trend over time

sales_date=sales_df.groupby('Date')['Amount'].sum()
print(sales_date)
```

```python
sales_trend=sales_df.groupby('Date')['Amount'].sum().reset_index()
```

```python
plt.plot(sales_trend['Date'], sales_trend['Amount'])
```

# 13.6 Data Visulaization with Seaborn

Seaborn is a Python visualization library based on Matplotlib that provides a high-level interface for drawing attractive and informative statistical graphics. Seaborn helps in creating complex visualizations with just a few lines of code. 

```python
import seaborn as sns
```

```python
tips = sns.load_dataset('tips')
tips
```

```python
sns.scatterplot(x='total_bill', y='tip', data=tips)
plt.title("Scatterplot of Total bills VS Tip")
```

```python
# Line plot

sns.lineplot(x='size', y='total_bill', data=tips )
```

```python
# Categorical variable

# Bar plot
sns.barplot(y='total_bill', x='day', data=tips )
```

```python
sns.barplot(y='total_bill', x='sex', data=tips )
```

```python
# boxplot

sns.boxplot(x='day', y='total_bill', data=tips)
```

```python
sns.boxplot(x='day', y='sex', data=tips)
```

```python
# violin plot
sns.violinplot(x='day', y='total_bill', data=tips)
```

```python
sns.violinplot(x='day', y='sex', data=tips)
```

```python
# histogram

sns.histplot(tips['total_bill'], bins=10, kde=True)
# KDE stands for Kernal Density Estimator, it smoothens the curvy.
```

```python
# KDE Plot

sns.kdeplot(tips['total_bill'], shade=True)
```

```python
# pair plot
sns.pairplot(tips)
```

```python
# heatmap

corr=tips[['total_bill', 'tip', 'size']].corr()
corr
```

```python
sns.heatmap(corr, annot=True)
```

```python
sns.heatmap(corr, annot=True, cmap='coolwarm')
```
