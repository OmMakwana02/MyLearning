# Section 15: Logging in Python

- Logging is a crucial aspect of any application, providing a way to track events, errors, and operational information. 
- Python's built-in module offers a flexible framework for emitting log messages from Python programs. 
- In this lesson, we weill cover the basics of logging, including how to configure logging, log levels, and best practices for ussing logging in Python applications.

## 15.1 Practical Impementation of Logging in Python

```python
import logging
```

```python
# Configure the basic logging settings. 
# There are many settings, eg. DEBUG, ERROR, etc...
logging.basicConfig(level=logging.DEBUG) 
# Debug is the initial basic logging settings that we need to do... 
```

```python
logging.basicConfig(level = logging.DEBUG) 
# Log messages

logging.debug("This is a debug message.")
logging.info("This is a info message.")
logging.warning("This is a warning message.")
logging.error("This is a error message")
logging.critical("This is a critical message.")
```

```python
# Configuring the logging...
logging.basicConfig(
    filename = 'app.log',
    filemode = 'w',
    level = logging.DEBUG,
    formate = '%(asctime)s-%(name)s-%(levelname)s-%(message)s',
    datefmt = '%Y-%m-%d %H:%M:%s'
    )

logging.debug("This is a debug message.")
logging.info("This is a info message.")
logging.warning("This is a warning message.")
logging.error("This is a error message")
logging.critical("This is a critical message.")
```
