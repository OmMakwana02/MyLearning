# Exception Handling
Exception Handling in Python allows you to handle errors gracefully and take corrective actions without stopping the execution of the program. This file covers the basic of exceptions, including how to use try, exceot, else and finally blocks.

## What is Exception??
Exceptions are events that disrupt the normal flow of program. They occur when an error is encountered during program execution. Common exceptions include:
- Divide by zero
- FileNotFoundError: File not found.
- ValueError: Invalid value.
- TypeError: Invalid type.

```python
# Eg.
a=b
# This will give a Name Error
```

```python
# To Solve the above block, we use Exception try, except block.
try:
    a=b 
except:
    print("The Variable is not been assigned")
```

```python
try:
    a=b 
except NameError as ex:
    print(ex)

# Here we handling the exception with the same type of the error class
```

```python
try:
    result = 1/0
except ZeroDivisionError as ex:
    print(ex)
    print("The denomnator is greator than 0")
```

```python
try:
    result = 1/2
    a=b
except ZeroDivisionError as ex:
    print(ex)
    print("The denomnator is greator than 0")

# Here, it will not catch the exception because we have derived the exception only for ZeroDivisionError class.
```

All the above mentioned exceptions are all derived from a single exception class "Exception".

```python
try:
    result = 1/2
    a=b
except ZeroDivisionError as ex:
    print(ex)
    print("The denomnator is greator than 0")
except Exception as ex1:
    print(ex1)
# We need make sure that the last exception must be this "Exception class", because its the parent class of all the exceptions.
```

```python
try:
    num = int(input("Enter a Number: "))
    result = 10/num
except ValueError:
    print("This is not a valid input")
except ZeroDivisionError:
    print("The Denominator must be greater than 0.")
except Exception as ex:
    print(ex)
```

```python
# try, except and else block
try:
    num = int(input("Enter a Number: "))
    result = 10/num
except ValueError:
    print("This is not a valid input")
except ZeroDivisionError:
    print("The Denominator must be greater than 0.")
except Exception as ex:
    print(ex)
else:
    print(f"The result is {result}")

# This else block makes sure that some part of code must bhi executed if there's no error in the code. If there will be any exception in the code then the except block will executed and it will come outside the try except block, but in case if the code has no error, then the else block has to be executed.
```

```python
# try, except, else and finally
try:
    num = int(input("Enter a Number: "))
    result = 10/num
except ValueError:
    print("This is not a valid input")
except ZeroDivisionError:
    print("The Denominator must be greater than 0.")
except Exception as ex:
    print(ex)
else:
    print(f"The result is {result}")
finally:
    print("Execution Complete.")

# If there is no error than only the else block will be exected, if there's an error then it will be handled by the except block. but in case of "finally" it will execute itself wether the error is there or not.
```
