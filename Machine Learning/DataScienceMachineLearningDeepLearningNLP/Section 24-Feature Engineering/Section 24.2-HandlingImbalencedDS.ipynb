## 24.2 Handling Imbalanced Dataset

- Wrt any problem statement that you are working in ML and DL, initially you will be given a data set. 
- So lets consider a dataset, and you are given a classification problem. Here, classification --> Supervised ML. 
- This means the output is basically in the form of categories.(If you have 2 categories than we can say it as a binary classification.)
- So lets condiser a Binary classification, lets say i have 1000 datapoints in my dataset.
- Out of these 1000 datapoints, 900 are 'Yes' datapoints and 100 are 'No' datapoints, so if we see the ratio of Yes:No its 9:1. This tells us that the dataset is imbalanced. 
- The main porblem with the imbalanced data set while you are solving a machine learning application, your model that you are probably creating, which will be responsible in predicting will get biased towards this maximum number of datapoints(900 yes).
- So its necessary that we fix this imbalanced data set, and we try to make sure that both the data points become equal.

## To Handle this, we use two techniques:

## 1. Up Sampling 
## 2. Down Sampling

```python
# Set the random seed for reproducibility
np.random.seed(123)

# Create a dataframe with two classes

n_sample = 1000
class_0_ratio = 0.9
n_class_0 = int(n_sample * class_0_ratio)
n_class_1 = n_sample - n_class_0
```

```python
n_class_0, n_class_1
```

```python
# Creating dataframe with Imbalanced datapoints

class_0 = pd.DataFrame({
    'feature_1': np.random.normal(loc=0, scale =1, size = n_class_0),
    'feature_2': np.random.normal(loc=0, scale =1, size = n_class_0),
    'target': [0] * n_class_0
})
class_1 = pd.DataFrame({
    'feature_1': np.random.normal(loc=2, scale =1, size = n_class_1),
    'feature_2': np.random.normal(loc=2, scale =1, size = n_class_1),
    'target': [1] * n_class_1
})
```

```python
df = pd.concat([class_0, class_1]).reset_index(drop=True)
```

```python
df
```

```python
df['target'].value_counts()
```

## --> Up Sampling

```python
# for rows with value as 1:
df_minority = df[df['target']==1]
# for rows with value as 0:
df_majority = df[df['target']==0]
```

```python
from sklearn.utils import resample
# This basically extrapolates the point: it means that whatever is the minority it will try to create new points for it, and it will try to equalize with respect to the majority.
```

```python
'''
Synatx: 
resample(minority parameter, replace = True, n_samples = len(majority), random_state = anyvalue)
arg 1: For which sample we want to probably performing upsampling.

arg 2: Here replace basically means that whenever I am performing the resampling over here, this sample will get added with the replacement.(So different different data points can be added over here.)

arg 3: How many we have to increase, here i will give the length of the majority population.

arg 4: for any random state value
'''
```

```python
df_minority_upsampling = resample(df_minority, 
                                  replace = True, 
                                  n_samples = len(df_majority), 
                                  random_state = 42)
```

```python
df_minority_upsampling.shape
```

```python
df_minority_upsampling.head()
```

```python
df_upsampled = pd.concat([df_majority, df_minority_upsampling])
df_upsampled
```

```python

df_upsampled['target'].value_counts()
```

## --> Down Sampling

```python
# Creating dataframe with Imbalanced datapoints

class_0 = pd.DataFrame({
    'feature_1': np.random.normal(loc=0, scale =1, size = n_class_0),
    'feature_2': np.random.normal(loc=0, scale =1, size = n_class_0),
    'target': [0] * n_class_0
})
class_1 = pd.DataFrame({
    'feature_1': np.random.normal(loc=2, scale =1, size = n_class_1),
    'feature_2': np.random.normal(loc=2, scale =1, size = n_class_1),
    'target': [1] * n_class_1
})
```

```python
dfd = pd.concat([class_0, class_1]).reset_index(drop=True)
```

```python
dfd['target'].value_counts()
```

```python
# for rows with value as 1:
dfd_minority = dfd[dfd['target']==1]
# for rows with value as 0:
dfd_majority = dfd[dfd['target']==0]
```

```python
dfd_majority_downsampling = resample(dfd_majority, 
                                  replace = False, 
                                  n_samples = len(dfd_minority), 
                                  random_state = 42)

''' 
Here in arg 2: we dont want ro replce data...we are reducing the data points
'''
```

```python
dfd_majority_downsampling.shape
```

```python
dfd_downsampled = pd.concat([dfd_minority, dfd_majority_downsampling])
dfd_downsampled
```

```python
dfd_downsampled.target.value_counts()
```
