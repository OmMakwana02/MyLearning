# Section 14 - SQL and SQLite
SQL (Structured Query Language) is a standard language for managing and manipulating relational databases. SQLite is a self-contained, serviceless, and xero-configuration database engine that is widely used for embedded database systems. In this lesson, we will cover the basics of SQL and SQLite, including creating databases, tables and performaing various SQL operations.

```python
import sqlite3
```

```python
# Connect to an SQLite database

connection = sqlite3.connect('example.db')
connection
```

```python
# Creating a cursor obj - to iterate through all th etables and rows

cursor = connection.cursor()
```

```python
# Creating a table
cursor.execute('''
Create Table If Not Exists employees(
    id Integer Primary Key,
    name Text Not Null,
    age Integer,
    department text
    )
''')

## Commit the changes
connection.commit()
```

```python
cursor.execute('''
Select * from employees
''')
```

```python
# Inserting the data in SQLite3

cursor.execute('''
insert into employees(name,age,department)
               values( 'Om', 20, 'Data Scientist')
''')

cursor.execute('''
insert into employees(name,age,department)
               values( 'Srishti', 23, 'AI Engineer')
''')

cursor.execute('''
insert into employees(name,age,department)
               values( 'Smith', 34, 'IT')
''')

cursor.execute('''
insert into employees(name,age,department)
               values( 'Samira', 25, 'Finance')
''')

connection.commit()
```

```python
# Query the data from the table
cursor.execute('Select * from employees')
rows = cursor.fetchall()

# print the queried data
for r in rows:
    print(r)
```

```python
# Update the data
cursor.execute('''
UPDATE employees
SET AGE=34
WHERE NAME="Om"
''')

connection.commit()
```

```python
cursor.execute('Select * from employees')
rows = cursor.fetchall()

# print the queried data
for r in rows:
    print(r)
```

```python
cursor.execute('''
DELETE FROM employees
WHERE id=44
''')

connection.commit()
```

```python
cursor.execute('Select * from employees')
rows = cursor.fetchall()

# print the queried data
for r in rows:
    print(r)
```

```python
cursor.execute('''
DELETE FROM employees
WHERE id=48
''')
cursor.execute('''
DELETE FROM employees
WHERE id=49
''')
cursor.execute('''
DELETE FROM employees
WHERE id=50
''')
cursor.execute('''
DELETE FROM employees
WHERE id=51
''')

connection.commit()
```

```python
cursor.execute('Select * from employees')
rows = cursor.fetchall()

# print the queried data
for r in rows:
    print(r)
```

```python
connection.close()
```

```python
connect = sqlite3.connect('practice.db')
connect
```

```python
cursor = connect.cursor()
```

```python
cursor.execute('''
CREATE TABLE IF NOT EXISTS sales_employee(
        ID int PRIMARY KEY,
        NAME TEXT NOT NULL,
        DATE text NOT NULL,
        AGE INT ,
        SALARY FLOAT NOT NULL
          )
''')
connect.commit()

data = [
    ('Om',      '20-02-2024',25,20000000),
    ('Srishti', '20-02-2024',23,15000000),
    ('Bhoomi',  '24-11-2019',18,5000000),
    ('Yashi',   '19-03-2003',30,364520),
    ('Hetvi',   '07-04-2012',35,453832)
]

# For bulk insertion
cursor.executemany('''
Insert into sales_employee(NAME, DATE, AGE, SALARY)
               values(?,?,?,?)
''',data)

connect.commit()
```

```python
cursor.execute('Select * from sales_employee')
rows = cursor.fetchall()

# print the queried data
for r in rows:
    print(r)
```

```python
connect.close()
```
