## 12.1 Iterators
Iterators are advanced Python concepts that allow for efficient looping and memory management. Iterators provide a way to access elements of a collection sequentially without exposing the underlying structure.

```python
my_list = [1,2,3,4,5,67,8,]
for i in my_list:
    print(i)
```

```python
'''
How to create iterator?
To create an Iterator we use "iter()" keyword.
'''
# Creating an iterator
iterator = iter(my_list)
print(type(iterator))
iterator
```

```python
'''
Here it uses lazy loadin technique, that basically means right now my "list_iterator" is created in some memory location, it will not give us any element usnless and until we iterate through that particular element. In order to iterate through all the elements we will use a new function "next()"
'''
next(iterator)

'''
So here as soon as we execute the next function, it will give you the 1st element of this iterator. As you go on executing this line of code, you will get the next element of the iterator. And at the last (means once the list is over) it will show you an error, that there are no elements in the list.
This next function by using lazy loading technique generates the memory over there. And by this we can see efficient looping and memory management.
'''
```

```python
iterator = iter(my_list)
```

```python
try:
    print(next(iterator))
except StopIteration:
    print("There are no elements in the iterator.")
```
